# Student Performance Analysis using Python

## Objective
The objective of this task is to analyze the student performance dataset using Python libraries such as Pandas, Matplotlib, and Seaborn.

The following operations are performed:
- Load the dataset
- Explore and clean the data
- Analyze student performance
- Create visualizations
- Draw conclusions from the dataset

```python
# Importing required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Step 1: Load Dataset

In this step, the dataset is loaded using Pandas.
The dataset contains student academic and personal information.

```python
# Loading dataset from local system

df = pd.read_csv(r"C:\Users\Dell\Downloads\student-mat.csv", sep=";")

# Display first 5 rows

df.head()
```

## Step 2: Inspect Dataset Shape

Checking the number of rows and columns present in the dataset.

```python
# Checking dataset shape

df.shape
```

## Step 3: Inspect Dataset Information

This step shows:
- Column names
- Data types
- Non-null values

```python
# Display dataset information

df.info()
```

## Step 4: Check Missing Values

Checking whether the dataset contains null or missing values.

```python
# Checking missing values

df.isnull().sum()
```

## Step 5: Check Duplicate Records

Checking duplicate rows in the dataset.

```python
# Checking duplicate values

df.duplicated().sum()
```

## Step 6: Remove Duplicate Records

Removing duplicate rows from the dataset for accurate analysis.

```python
# Removing duplicate rows

df = df.drop_duplicates()

# Checking updated shape

df.shape
```

## Step 7: Calculate Average Final Grade (G3)

Calculating the average value of final grades scored by students.

```python
# Calculating average final grade

average_g3 = df["G3"].mean()

print("Average Final Grade (G3):", average_g3)
```

## Step 8: Students Scoring Above 15

Counting students whose final grade is greater than 15.

```python
# Counting students scoring above 15

students_above_15 = (df["G3"] > 15).sum()

print("Number of students scoring above 15:", students_above_15)
```

## Step 9: Correlation Between Study Time and Performance

Analyzing whether study time affects final grades.
A positive correlation means students with higher study time generally score better.

```python
# Finding correlation between study time and final grade

correlation = df["studytime"].corr(df["G3"])

print("Correlation between study time and performance:", correlation)
```

## Step 10: Gender-wise Performance Analysis

Comparing average final grades of male and female students.

```python
# Calculating average final grades based on gender

gender_performance = df.groupby("sex")["G3"].mean()

print(gender_performance)
```

## Step 11: Histogram of Final Grades

Creating a histogram to understand grade distribution among students.

```python
# Creating histogram of final grades

plt.figure(figsize=(8,5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Histogram of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Frequency")

plt.show()
```

## Step 12: Scatterplot - Study Time vs Final Grade

Creating a scatterplot to visualize the relationship between study time and student performance.

```python
# Creating scatterplot for study time and grades

plt.figure(figsize=(8,5))

plt.scatter(df["studytime"], df["G3"], alpha=0.7)

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

## Step 13: Bar Chart - Male vs Female Average Score

Creating a bar chart to compare average final grades of male and female students.

```python
# Creating bar chart for gender-wise performance

gender_performance.plot(kind="bar", figsize=(6,4), color=["blue", "pink"])

plt.title("Male vs Female Average Final Score")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade (G3)")

plt.xticks(rotation=0)

plt.show()
```
